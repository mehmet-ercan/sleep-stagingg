```python
import pyedflib
import numpy as np

edf_file_path = "/mnt/data1/data/owndata/3.to-mongo-examples/6b19b899-e4b2-4465-93d7-ab2c329b092d/Traces.edf"

target_channel_name = "Activity"  # veya indeks için: target_channel_index = 0
output_txt_path = "6b19b899-e4b2-4465-93d7-ab2c329b092d_Activity-fromfile.txt"

with pyedflib.EdfReader(edf_file_path) as f:
    channel_labels = f.getSignalLabels()
    print("Mevcut Kanallar:", channel_labels)

    try:
        channel_index = channel_labels.index(target_channel_name)
    except ValueError:
        raise Exception(f"Kanal '{target_channel_name}' bulunamadı. Yukarıdaki kanallardan birini seçin.")

    signal = f.readSignal(channel_index).astype(np.float64)

with open(output_txt_path, "w") as txt_file:
    for value in signal:
        txt_file.write(f"{value:.18e}\n")

print(f"'{target_channel_name}' kanalına ait veriler float64 olarak '{output_txt_path}' dosyasına yazıldı.")
```
